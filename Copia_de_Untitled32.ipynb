## Cálculo del Área Bajo la Curva

Vamos a calcular el área bajo la curva de la función $f(x) = x^4 - 2x^3 + 3x + 5$ en el intervalo $[0, 3]$. Primero, definiremos la función y luego procederemos con el cálculo exacto usando integración simbólica y aproximaciones numéricas (sumas de Riemann).

### 1. Definición de la Función

```pseudocode
# PSEUDOCÓDIGO para definir la función f(x)

# Creamos una función 'f' que toma 'x'.
# Calcula x^4 - 2x^3 + 3x + 5 y devuelve ese resultado.
# Luego, imprimimos la fórmula de la función.
```

````python
def f(x):
    return x**4 - 2*x**3 + 3*x + 5

print("Función definida: f(x) = x^4 - 2x^3 + 3x + 5")
````

### 2. Cálculo Exacto del Área (Integral Definida) con SymPy

Utilizaremos la librería `SymPy` para realizar el cálculo simbólico de la integral definida. Esto nos dará el valor exacto del área.

```pseudocode
# PSEUDOCÓDIGO para el cálculo exacto del área con SymPy

# Instalamos y cargamos la librería SymPy.
# Definimos 'x' como un símbolo para SymPy.
# Escribimos la función con la 'x' simbólica.
# Calculamos la integral de la función entre 0 y 3 para el 'área exacta'.
# Mostramos la función simbólica y el área exacta.
```

````python
# Instalar SymPy si no está instalado
!pip install sympy

import sympy
from sympy import integrate, Symbol

x = Symbol('x')

# Definir la función simbólica
f_sym = x**4 - 2*x**3 + 3*x + 5

# Calcular la integral definida en el intervalo [0, 3]
exact_area = integrate(f_sym, (x, 0, 3))

print(f"La función simbólica es: {f_sym}")
print(f"El área exacta bajo la curva en [0,3] es: {exact_area}")
````

### 3. Cálculo Aproximado del Área (Sumas de Riemann)

Ahora, calcularemos el área por defecto (suma inferior) y por exceso (suma superior) utilizando sumas de Riemann. Dividiremos el intervalo $[0, 3]$ en un número `n` de subintervalos. Cuanto mayor sea `n`, mejor será la aproximación.

**Suma Inferior (por Defecto):** Se utilizan las alturas mínimas de la función en cada subintervalo.

**Suma Superior (por Exceso):** Se utilizan las alturas máximas de la función en cada subintervalo.

```pseudocode
# PSEUDOCÓDIGO para Sumas de Riemann (por defecto, por exceso y punto medio)

# Definimos el intervalo [a, b] y el número de divisiones 'n'.
# Calculamos el ancho de cada división 'delta_x'.

# Inicializamos 'lower_sum' y 'upper_sum' a cero.
# Para cada división:
#   Calculamos los valores de la función en los extremos del pedacito.
#   A 'lower_sum' le sumamos el mínimo valor * delta_x.
#   A 'upper_sum' le sumamos el máximo valor * delta_x.

# Mostramos las sumas inferior y superior.

# Inicializamos 'midpoint_sum' a cero.
# Para cada división:
#   Calculamos el punto medio del pedacito.
#   A 'midpoint_sum' le sumamos el valor de la función en el punto medio * delta_x.
# Mostramos la suma del punto medio.
```

````python
# Definir el intervalo y el número de subintervalos
a = 0
b = 3
n = 200 # Número de subintervalos para la aproximación

# Calcular el ancho de cada subintervalo (delta_x)
delta_x = (b - a) / n

# Inicializar las sumas
lower_sum = 0
upper_sum = 0

# Calcular las sumas de Riemann
for i in range(n):
    x_i = a + i * delta_x
    x_i_plus_1 = a + (i + 1) * delta_x

    val_at_left = f(x_i)
    val_at_right = f(x_i_plus_1)

    lower_sum += min(val_at_left, val_at_right) * delta_x
    upper_sum += max(val_at_left, val_at_right) * delta_x


print(f"Número de subintervalos (n): {n}")
print(f"Ancho del subintervalo (delta_x): {delta_x:.4f}")
print(f"Área aproximada por defecto (Suma Inferior): {lower_sum:.4f}")
print(f"Área aproximada por exceso (Suma Superior): {upper_sum:.4f}")

# También podemos calcular la suma de Riemann del punto medio como una buena aproximación
midpoint_sum = 0
for i in range(n):
    mid_point = a + (i + 0.5) * delta_x
    midpoint_sum += f(mid_point) * delta_x

print(f"Área aproximada por punto medio: {midpoint_sum:.4f}")
````

### 5. Método del Punto Medio (Revisado con M y Cota de Error)

El método del punto medio usa el valor de la función en el punto medio de cada subintervalo para aproximar el área. La cota de error para este método depende del máximo de la segunda derivada de la función, $M_2 = \max |f''(x)|$ en el intervalo $[a,b]$.

**Cota de Error:** $E_M \le \frac{M_2(b-a)^3}{24n^2}$

```pseudocode
# PSEUDOCÓDIGO para Método del Punto Medio (con 'M' y cota de error)

# Calculamos la segunda derivada de la función con SymPy.
# Encontramos 'M2', el valor máximo de la segunda derivada en el intervalo.
# Usamos el 'midpoint_sum' que ya calculamos como área.
# Calculamos la cota de error para el punto medio usando 'M2'.
# Mostramos M2, el área del punto medio y su cota de error.
```

````python
from sympy import diff, Abs

# Segunda derivada de la función
f_double_prime = diff(f_sym, x, 2)
print(f"Segunda derivada f''(x): {f_double_prime}")

# Encontrar el máximo de la segunda derivada absoluta en [a, b]
# Para f''(x) = 12x^2 - 12x, los puntos críticos de |f''(x)| pueden estar en x=0, x=1 (raíces) y los extremos del intervalo.
# Evaluamos |f''(x)| en los extremos del intervalo y en puntos donde f'''(x)=0

f_triple_prime = diff(f_sym, x, 3)
# Puntos donde f'''(x) = 0: 24x - 12 = 0 => x = 0.5

points_to_check = [a, b, 1/2]
points_to_check = [p for p in points_to_check if a <= p <= b]

M2_values = [Abs(f_double_prime.subs(x, p)) for p in points_to_check]

M2 = max(M2_values)

# Si M2_values está vacío (ej. si el intervalo no contiene 0.5), toma los extremos
if not M2_values and a <= b: # Ensure a valid interval
    M2 = max(Abs(f_double_prime.subs(x, a)), Abs(f_double_prime.subs(x, b)))


# Cálculo del área por el método del punto medio (ya calculado como midpoint_sum)
midpoint_area = midpoint_sum

# Cota de error del método del punto medio
error_bound_midpoint = (M2 * (b - a)**3) / (24 * n**2)

print(f"Valor de M2 (máx |f''(x)|) para el método del punto medio: {M2:.4f}")
print(f"Área por el método del punto medio: {midpoint_area:.4f}")
print(f"Cota de error para el método del punto medio: {error_bound_midpoint:.6f}")
````

### 6. Método Trapezoidal (Revisado con M y Cota de Error)

El método trapezoidal aproxima el área usando trapezoides en lugar de rectángulos. La cota de error también depende del máximo de la segunda derivada de la función, $M_2 = \max |f''(x)|$ en el intervalo $[a,b]$.

**Cota de Error:** $E_T \le \frac{M_2(b-a)^3}{12n^2}$

```pseudocode
# PSEUDOCÓDIGO para Método Trapezoidal (con 'M' y cota de error)

# Reutilizamos 'M2' del método del punto medio.
# Inicializamos 'trapezoidal_area' a cero.
# Para cada división:
#   Sumamos el área del trapecio (promedio de f en los extremos * delta_x).
# Calculamos la cota de error trapezoidal usando 'M2'.
# Mostramos M2, el área trapezoidal y su cota de error.
```

````python
# El valor de M2 es el mismo que para el método del punto medio
M2_trap = M2 # Reutilizamos el M2 calculado anteriormente

# Cálculo del área por el método trapezoidal
trapezoidal_area = 0
for i in range(n):
    x_i = a + i * delta_x
    x_i_plus_1 = a + (i + 1) * delta_x
    trapezoidal_area += (f(x_i) + f(x_i_plus_1)) / 2 * delta_x

# Cota de error del método trapezoidal
error_bound_trapezoidal = (M2_trap * (b - a)**3) / (12 * n**2)

print(f"Valor de M2 (máx |f''(x)|) para el método trapezoidal: {M2_trap:.4f}")
print(f"Área por el método trapezoidal: {trapezoidal_area:.4f}")
print(f"Cota de error para el método trapezoidal: {error_bound_trapezoidal:.6f}")
````

### 7. Método de Simpson (Revisado con M y Cota de Error)

El método de Simpson utiliza parábolas para aproximar el área, lo que generalmente ofrece una mayor precisión. Requiere que el número de subintervalos `n` sea par. La cota de error depende del máximo de la cuarta derivada de la función, $M_4 = \max |f^{(4)}(x)|$ en el intervalo $[a,b]$.

**Cota de Error:** $E_S \le \frac{M_4(b-a)^5}{180n^4}$

```pseudocode
# PSEUDOCÓDIGO para Método de Simpson (con 'M' y cota de error)

# Verificamos que 'n' sea par (Simpson lo necesita).
# Calculamos la cuarta derivada de la función con SymPy.
# Encontramos 'M4', el valor máximo de la cuarta derivada (para nuestra función es 24).
# Inicializamos 'simpson_area' (usando la fórmula de Simpson con pesos 1, 4, 2, 4, ..., 2, 4, 1).
# Calculamos la cota de error de Simpson usando 'M4'.
# Mostramos M4, el área de Simpson y su cota de error.
```

````python
# Verificar que n sea par para el método de Simpson
if n % 2 != 0:
    print("Advertencia: El método de Simpson requiere un número par de subintervalos (n).")
    # Si n fuera impar y se necesitara ajustar, se podría hacer aquí
    # Por simplicidad, asumimos que n ya es par (como 200 en este caso)

# Cuarta derivada de la función
f_fourth_prime = diff(f_sym, x, 4)
print(f"Cuarta derivada f''''(x): {f_fourth_prime}")

# Encontrar el máximo de la cuarta derivada absoluta en [a, b]
# Para f''''(x) = 24 (una constante), el máximo absoluto es simplemente 24.
M4 = Abs(f_fourth_prime.subs(x, a)) # Cualquier punto en [a,b] dará el mismo valor

# Cálculo del área por el método de Simpson
simpson_area = f(a) + f(b)
for i in range(1, n):
    x_i = a + i * delta_x
    if i % 2 == 0:
        simpson_area += 2 * f(x_i)
    else:
        simpson_area += 4 * f(x_i)

simpson_area = simpson_area * (delta_x / 3)

# Cota de error del método de Simpson
error_bound_simpson = (M4 * (b - a)**5) / (180 * n**4)

print(f"Valor de M4 (máx |f''''(x)|) para el método de Simpson: {M4:.4f}")
print(f"Área por el método de Simpson: {simpson_area:.4f}")
print(f"Cota de error para el método de Simpson: {error_bound_simpson:.6f}")
````
